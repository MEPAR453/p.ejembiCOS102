```python
import tkinter as tk
from tkinter import ttk

# Parent Class
class Zenith:
    def mutual_services(self):
        return [
            "Lines of credit",
            "Investment management and accounts",
            "Insurance"
        ]

    def unique_services(self):
        return []

# Child Classes
class RetailBanking(Zenith):
    def unique_services(self):
        return [
            "Retirement and education accounts",
            "Loans and mortgages",
            "Checking and savings"
        ]

class GlobalBanking(Zenith):
    def unique_services(self):
        return [
            "Multi-currency management",
            "Foreign currency accounts",
            "Foreign currency credit cards",
            "Transborder advisory services",
            "Liquidity management"
        ]

class CommercialBanking(Zenith):
    def unique_services(self):
        return [
            "Advisory services"
        ]

# GUI Application
class App:
    def __init__(self, root):
        self.root = root
        self.root.title("Zenith Bank Services")

        self.main_frame = tk.Frame(root)
        self.main_frame.pack(expand=True)

        # Labels and Inputs
        tk.Label(self.main_frame, text="Employee Name").grid(row=0, column=0, pady=10)
        self.name_entry = tk.Entry(self.main_frame)
        self.name_entry.grid(row=0, column=1, pady=10)

        tk.Label(self.main_frame, text="Division").grid(row=1, column=0, pady=10)

        self.division = ttk.Combobox(self.main_frame, values=[
            "Retail", "Global", "Commercial"
        ])
        self.division.grid(row=1, column=1, pady=10)

        # Button
        tk.Button(self.main_frame, text="Show Services", command=self.show_services, fg="white", bg="black").grid(row=2, column=0, columnspan=2,)

        # Output
        self.output = tk.Text(self.main_frame, height=15, width=50)
        self.output.grid(row=3, column=0, columnspan=2)

    def show_services(self):
        name = self.name_entry.get()
        division = self.division.get()

        if division == "Retail":
            obj = RetailBanking()
        elif division == "Global":
            obj = GlobalBanking()
        elif division == "Commercial":
            obj = CommercialBanking()
        else:
            self.output.insert(tk.END, "Invalid Division\n")
            return

        # Clear output
        self.output.delete(1.0, tk.END)

        # Display results
        self.output.insert(tk.END, f"Employee: {name}\n")
        self.output.insert(tk.END, f"Division: {division}\n\n")

        self.output.insert(tk.END, "Mutual Services:\n")
        for service in obj.mutual_services():
            self.output.insert(tk.END, f"- {service}\n")

        self.output.insert(tk.END, "\nUnique Services:\n")
        for service in obj.unique_services():
            self.output.insert(tk.END, f"- {service}\n")


# Run App
root = tk.Tk()
app = App(root)
root.mainloop()
```
