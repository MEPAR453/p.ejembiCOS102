```python
import tkinter as tk
from tkinter import ttk
from tkinter import messagebox
from abc import ABC, abstractmethod

#OOP BACKEND (Abstract Class & Subclasses)
class External_Vendors(ABC):
   
    @abstractmethod
    def menu(self):
        pass

class Cooperative_Cafeteria(External_Vendors):
    def menu(self):
        return (
            "--- Cooperative Cafeteria Menu ---\n"
            "Jollof Rice and Stew  - ₦200\n"
            "White Rice and Stew   - ₦200\n"
            "Fried Rice            - ₦200\n"
            "Salad                 - ₦100\n"
            "Plantain              - ₦100"
        )

class Faith_Hostel_Cafeteria(External_Vendors):
    def menu(self):
        return (
            "--- Faith Hostel Cafeteria Menu ---\n"
            "Fried Rice            - ₦400\n"
            "White Rice and Stew   - ₦400\n"
            "Jollof Rice           - ₦400\n"
            "Beans                 - ₦200\n"
            "Chicken               - ₦1000"
        )

class Student_Centre_Cafeteria(External_Vendors):
    def menu(self):
        return (
            "--- Student Centre Cafeteria Menu ---\n"
            "Chicken Fried Rice    - ₦800\n"
            "Pomo Sauce            - ₦300\n"
            "Spaghetti Jollof      - ₦500\n"
            "Amala/Ewedu           - ₦500\n"
            "Semo with Eforiro Soup - ₦500"
        )


#GUI FRONTEND (Tkinter Interface)
class VendorManagerGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("PAU External Food Vendors Management")
        self.root.geometry("400x400")
        self.root.configure(bg="#f4f4f4")
        
        # Heading Label
        self.title_label = tk.Label(
            root, text="PAU Service Unit Portal", 
            font=("Arial", 16, "bold"), bg="#f4f4f4", fg="#333"
        )
        self.title_label.pack(pady=15)
        
        # Dropdown Label
        self.select_label = tk.Label(
            root, text="Select External Vendor of Interest:", 
            font=("Arial", 11), bg="#f4f4f4"
        )
        self.select_label.pack(pady=5)
        
        # Combobox / Dropdown for selection
        self.vendor_options = [
            "Cooperative Cafeteria", 
            "Faith Hostel Cafeteria", 
            "Student Centre Cafeteria"
        ]
        self.dropdown = ttk.Combobox(root, values=self.vendor_options, state="readonly", width=30, font=("Arial", 10))
        self.dropdown.pack(pady=5)
        self.dropdown.set("Choose a vendor...") # Placeholder text
        
        # Display Menu Button
        self.btn_display = tk.Button(
            root, text="Display Menu", command=self.fetch_and_display_menu,
            bg="#2c3e50", fg="white", font=("Arial", 10, "bold"), padx=10, pady=5
        )
        self.btn_display.pack(pady=15)
        
        # Text Box Area to Output the Menu results
        self.output_text = tk.Text(root, height=10, width=45, font=("Courier", 11), bg="#fff", bd=2, relief="groove")
        self.output_text.pack(pady=10)
        
    def fetch_and_display_menu(self):
        # Clear previous text output
        self.output_text.delete("1.0", tk.END)
        
        selected_vendor = self.dropdown.get()
        
        # Polymorphic initialization based on dropdown selection
        if selected_vendor == "Cooperative Cafeteria":
            vendor_instance = Cooperative_Cafeteria()
        elif selected_vendor == "Faith Hostel Cafeteria":
            vendor_instance = Faith_Hostel_Cafeteria()
        elif selected_vendor == "Student Centre Cafeteria":
            vendor_instance = Student_Centre_Cafeteria()
        else:
            messagebox.showwarning("Selection Error", "Please select a valid vendor from the dropdown list.")
            return

        # Invoke the menu() method polymorphically and display it in the GUI box
        menu_details = vendor_instance.menu()
        self.output_text.insert(tk.END, menu_details)

#App Execution
if __name__ == "__main__":
    root = tk.Tk()
    app = VendorManagerGUI(root)
    root.mainloop()
```
