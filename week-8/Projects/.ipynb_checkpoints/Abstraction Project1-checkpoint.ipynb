```python
from abc import ABC, abstractmethod

# 1. Define the Abstract Base Class
class coup_de_ecriva(ABC):
    
    @abstractmethod
    def Fan_Page(self):
        pass

# 2. Define the Subclasses using Single Inheritance
class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to FC Cirok")

class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to Madiba FC")

class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to Blue Jay FC")

class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print("Welcome to TSG Walker")


# 3. Main Program for User Input and Dynamic Instantiation
def main():
    print("--- Welcome to Coup de Ecriva ---")
    print("Please select the club you support:")
    print("1. FC Cirok")
    print("2. Madiba FC")
    print("3. Blue Jay FC")
    print("4. TSG Walker")
    
    user_choice = input("\nEnter the name or number of your club: ").strip().lower()
    
    # Map the user's input to the corresponding class object
    if user_choice in ["1", "fc cirok", "fc_cirok"]:
        club_object = FC_Cirok()
    elif user_choice in ["2", "madiba fc", "madiba_fc"]:
        club_object = Madiba_FC()
    elif user_choice in ["3", "blue jay fc", "blue_jay_fc"]:
        club_object = Blue_Jay_FC()
    elif user_choice in ["4", "tsg walker", "tsg_walker"]:
        club_object = TSG_Walker()
    else:
        print("Invalid choice. Club not found.")
        return # Exit the function early if no valid club is picked

    # Invoke the Fan_Page method from the instantiated subclass object
    print("\n--- Fan Page Output ---")
    club_object.Fan_Page()

if __name__ == "__main__":
    main()
```
