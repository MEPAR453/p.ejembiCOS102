```python
import tkinter as tk
import customtkinter as ctk

ctk.set_appearance_mode("dark")
ctk.set_default_color_theme("blue")

class Signup_page(ctk.CTk):
    def __init__(self, root):
        self.root = root
        self.root.title ("Sign-up Page")
        self.geometry("1152x720")
        
        self.main_frame = ctk.CTkFrame(root, fg_colour="#171923")
        self.main_frame.pack(fill= "both", expand=True)

        #Login frame
        self.login_frame = ctk.CTkFrame(self.main_frame, width=491, height=568, 
                                        corner_radius=20, fg_colour="#282C3E")
        self.login-frame.place(relx=0.5,rely=0.5, anchor="center")

        # Logo
        self.logo = ctk.CTkLabel(self.main_frame, text="KULTURE", font=("Black Italic", 96, "bold"),
            text_color="#DC2F02")

        self.login_text.pack(pady=(25, 30))

        # Username
        self.username_label = ctk.CTkLabel(self.login_frame, text="Username", 
                                           font=("Black Italic", 40, "bold"),text_colour="#FFFFFF")
        
        self.username_label.pack(anchor="w", padx=40)

        self.username_entry = ctk.CTkEntry(self.login_frame, width=415, height=59, corner_radius=20)

        self.username_entry.pack(pady=10)

        self.password_label = ctk.CTkLabel(self.login_frame, text="Password", 
                                           font=("Black Italic", 40, "bold"),text_colour="#FFFFFF")

        self.password_label.pack(anchor="w", padx=40, pady=(10, 0))

        self.password_entry = ctk.CTkEntry(self.login_frame, width=415, height=59, corner_radius=20,show="*")
        
        self.password_entry.pack(pady=10)

        # Button
        self.enter_button = ctk.CTkButton(self.login_frame,text="Enter", width=240, height=79,
                                          corner_radius=35,fg_color="#FFFFFF", hover_color="#171923",
                                         font=("Black Italic", 24, "bold"), text_colour="#FFFFFF")

        self.enter_button.pack(pady=25)


# Run App
root = ctk.CTk()
app = SignupPage(root)
root.mainloop()
                        

       
```
