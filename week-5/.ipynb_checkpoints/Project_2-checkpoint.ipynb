```python
import tkinter as tk
from tkinter import ttk

# 1. The Delivery Class (The OOP Logic)
class Delivery:
    def calculate_charge(self, location, weight):
        # Pricing logic for PAU
        if location == "PAU":
            if weight >= 10:
                return 2000
            else:
                return 1500
        
        # Pricing logic for Epe
        elif location == "Epe":
            if weight >= 10:
                return 5000
            else:
                return 4000
        
        return 0

#GUI Functionality
def process_calculation():
    location = combo_location.get()
    weight_str = entry_weight.get()

    try:
        weight = float(weight_str)
        system = Delivery()
        charge = system.calculate_charge(location, weight)

        # Update the display box
        result_box.config(state='normal')
        result_box.delete('1.0', tk.END)
        
        output = (f"--- DELIVERY INVOICE ---\n"
                  f"Location: {location}\n"
                  f"Weight: {weight}kg\n"
                  f"------------------------\n"
                  f"Total Charge: ₦{charge}")
        
        result_box.insert(tk.END, output)
        result_box.config(fg="black")
        result_box.config(state='disabled')

    except ValueError:
        # Error handling for non-numeric weight input
        result_box.config(state='normal')
        result_box.delete('1.0', tk.END)
        result_box.insert(tk.END, "ERROR: Please enter a valid number for weight.")
        result_box.config(fg="red")
        result_box.config(state='disabled')

#GUI Layout Setup
root = tk.Tk()
root.title("Logistics Price Calculator")
root.geometry("400x450")

#Labels and Entries
# Header at the top
header_label = tk.Label(root, text="Logistics Price Calculator",
                        font=("Helvetica", 16, "bold"), fg="white", bg="dark red", pady=10)
header_label.pack(fill="x")

#Header
tk.Label(root, text="Package Delivery Service", font=("Helvetica", 14, "bold")).pack(pady=10)

#Location Input
tk.Label(root, text="Select Destination:", font=("Arial", 10)).pack(pady=5)
combo_location = ttk.Combobox(root, values=["PAU", "Epe"], state="readonly")
combo_location.pack(pady=5)
combo_location.set("PAU") # Default selection

#Weight Input
tk.Label(root, text="Enter Package Weight (kg):", font=("Arial", 10)).pack(pady=5)
entry_weight = tk.Entry(root, font=("Arial", 12), width=15)
entry_weight.pack(pady=5)

#Calculate Button
btn_calculate = tk.Button(root, text="Calculate Price", command=process_calculation, 
                          bg="green", fg="white", font=("Arial", 10, "bold"), padx=20)
btn_calculate.pack(pady=20)

#Output Text Box
tk.Label(root, text="Response:", font=("Arial", 9)).pack()
result_box = tk.Text(root, height=8, width=40, font=("Courier", 10), 
                     state='disabled', bg="#eeeeee")
result_box.pack(pady=10)

root.mainloop()
```
