```python
import tkinter as tk
from tkinter import messagebox as msgbox
from tkinter import ttk
import random

Employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu", 
            "Stella Mankinde", "Jane Akibo", "Ago James", "Michell Taiwo", 
            "Abraham Jones", "Nicole Anide", "Kosi Korso", "Adele Martins", 
            "Emmanuel Ojo", "Ajayi Fatima"]

Tasks = ["Loading", "Transporting", "Reviewing Orders", "Customer Service", "Delivering Items"]

#The Employee Class
class Employee:
    def check_employee(self, name):
        return name.lower() in [emp.lower() for emp in Employees]

    def take_attendance(self, name):
        return f" attendance marked for {name.title()}."

    def assign_task(self):
        index = random.randint(0, len(Tasks) -1) #uses random.randint as requested in the hint
        return Tasks[index]

    def refuse_access(self):
        #Logic for a failed login
        return "Access Denied: You are not on the employee list."
        
#GUI Setup and Functions
def process_login():
    #Get name from the entry box
    user_input = entry_name.get().strip() 
    system = Employee()   #Create an instance of our class
    
    result_box.config(state='normal')
    result_box.delete('1.0', tk.END)
    
    if system.check_employee(user_input):
        attendance_msg = system.take_attendance(user_input)
        task_msg = system.assign_task()
        
        #Insert result into the text box
        final_output = f"SUCCESS:\n{attendance_msg}\nYour task for today: {task_msg}"
        result_box.insert(tk.END, final_output)
        result_box.config(fg="green") 
    else:
        error_msg = system.refuse_access()
        result_box.insert(tk.END, error_msg)
        result_box.config(fg="red") 
    
    result_box.config(state='disabled') 


#creating the Window
root = tk.Tk()
root.title("Jane's Logistics Employee management System")
root.geometry("400x400")

#Labels and Entries
header_label = tk.Label(root, text="Jane's Logistics",
                        font=("Helvetica", 16, "bold"), fg="white", bg="purple", pady=10)
header_label.pack(fill="x")
header_label = tk.Label(root, text="Employee management system",
                        font=("Helvetica", 10,"bold"), fg="white", bg="red", pady=10)
header_label.pack(fill="x")

# Big frame for inputs
# Title above frame
# Normal frame (no text)
big_frame = tk.Frame(root, padx=20, pady=20, relief="groove", bd=2)
big_frame.pack(padx=20, pady=30, fill="x")

# Widgets
tk.Label(root, text="Enter Employee Name:", font=("Arial", 12)).pack(pady=10)
entry_name = tk.Entry(root, font=("Arial", 12), width=30)
entry_name.pack(pady=10)

btn_verify = tk.Button(text="Verify Identity", command=process_login, 
                       bg="blue", fg="white", font=("Arial", 10, "bold"), width=15)
btn_verify.pack(padx=5)

tk.Label(root, text="System Response:", font=("Arial", 10, "italic")).pack(pady=(10, 0))
result_box = tk.Text(root, height=6, width=45, font=("Arial", 10), 
                     state='disabled', bg="#f0f0f0", padx=10, pady=10)
result_box.pack(pady = 10)

root.mainloop()
```
